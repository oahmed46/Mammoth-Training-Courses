```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
```

```python
img = Image.open("image.jpg")
plt.imshow(img)
```

```python
rgb_array = np.array(img)
print(rgb_array)
```

```python
formula = [0.2989, 0.5870, 0.1140]
```

```python
greyscale = rgb_array[..., :3] @ formula
```

```python
grey_image = Image.fromarray(greyscale)
plt.imshow(grey_image)
```

```python
greyscale_2 = np.dot(rgb_array[..., :3], formula)
```

```python
grey_image2 = Image.fromarray(greyscale_2)
plt.imshow(grey_image2)
```
