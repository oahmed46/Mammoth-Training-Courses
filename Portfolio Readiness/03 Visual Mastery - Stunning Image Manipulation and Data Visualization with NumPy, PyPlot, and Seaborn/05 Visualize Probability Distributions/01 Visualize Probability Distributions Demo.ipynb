```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
samples = [1,10,50,100,500,1000]
smp_means = []
```

```python
for s in samples:
    np.random.seed(1)
    x = [np.mean(np.random.randint(-40,40,s)) for _i in range(1000)]
    smp_means.append(x)
print(smp_means)
```

```python
fig, ax = plt.subplots(3,2, figsize=(8,8))
k = 0
for i in range(0,3):
    for j in range(0,2):
        ax[i,j].hist(smp_means[k],10, density = True)
        ax[i, j].set_title(label=samples[k])
        k += 1
plt.show()
```

```python
smp_means_exp = []
for j in samples:
    np.random.seed(1)
    x = [np.mean(np.random.exponential(scale=40,size=j)) for _i in range(1000)]
    smp_means_exp.append(x)
print(smp_means_exp)
```

```python
fig, ax = plt.subplots(3,2, figsize=(8,8))
idx = 0
for i in range(0,3):
    for j in range(0,2):
        ax[i,j].hist(smp_means_exp[idx],10,density=True)
        ax[i,j].set_title(label=samples[idx])
        idx += 1
plt.show()
```
