```python
import cv2
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
```

```python
img = Image.open("image.jpg")
plt.imshow(img)
```

```python
img_array = np.array(img)
print(img_array)
```

```python
img_array = np.array(img)/255
print(img_array)
```

```python
image_blur = cv2.blur(img_array, (15,15))
```

```python
plt.imshow(image_blur)
```

```python
image_blur2 = cv2.blur(img_array, (3,3))
plt.imshow(image_blur2)
```

```python
image_blur3 = cv2.blur(img_array, (10,10))
plt.imshow(image_blur3)
```

```python
image_blur4 = cv2.blur(img_array, (20,20))
plt.imshow(image_blur4)
```

```python
image_gaussianblur = cv2.GaussianBlur(img_array, (15,15), 0)
plt.imshow(image_gaussianblur)
```
