```python
con_matrix = [[13,5,5,3],
              [11,0,21,13],
              [4,-10,34,2],
              [3,5,25,7]]

Var1 = ["V1-1","V1-2","V1-3","V1-4"]
Var2 = ["V2-1","V2-2","V2-3","V2-4"]

con_matrix_df = pd.DataFrame(con_matrix, columns=Var2, index=Var1)

plt.figure(figsize=(10,10))
sns.set(font_scale=1.5)
sns.heatmap(con_matrix_df,
            cmap="coolwarm",
            annot=True,
            linewidths=1)

plt.xlabel("Variable 2", fontsize=22)
plt.xlabel("Variable 1", fontsize=22)
```

```python
con_matrix = [[4963,5,5,3,3,3],
              [11,4842,21,2,5,5],
              [4,1,4999,2,6,13],
              [3,5,25,4825,30,96],
              [1,4,15,10,4873,124],
              [2,4,18,18,84,4943]]

lang = ["deu","eng","fra","ita","por","spa"]
con_matrix_df = pd.DataFrame(con_matrix, columns=lang, index=lang)

plt.figure(figsize=(10,10))
sns.set(font_scale=1.5)

sns.heatmap(con_matrix_df,
            cmap="coolwarm",
            annot=True,
            fmt="0.5g",
            vmax=200)

plt.xlabel("Predicted", fontsize=22)
plt.ylabel("Actual", fontsize=22)
```
