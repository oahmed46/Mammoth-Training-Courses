```python
def find_max_array(input,k):
    max_sum, window_sum = 0,0
    start_index = 0

    for element in range(len(input)):
        
        window_sum += input[element]

        if element + 1 >= k:

            max_sum = max(max_sum, window_sum)

            window_sum -= input[start_index]

            start_index += 1

    return max_sum

find_max_array([1,4,8,2],2)
```
