```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
#from google.colab.patches import cv2_imshow
```

```python
# img = cv2.imread("img.jpg")
# cv2_imshow(img)
img = cv2.imread("img.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis("off")
```

```python
# grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# cv2_imshow(grayscale_img)
grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
grayscale_img = cv2.cvtColor(grayscale_img, cv2.COLOR_BGR2RGB)
plt.imshow(grayscale_img)
plt.axis("off")
```

```python
grayscale_img.shape
```

```python
rgb_array = np.array(img)
```

```python
# Optional manual way to grayscale
grayscale2 = rgb_array[...,:3] @ [0.2989, 0.5870, 0.1140]
print(grayscale2)
#cv2_imshow(grayscale2)
```

```python
grayscale2_uint8 = grayscale2.astype(np.uint8)
grayscale2_rgb = cv2.cvtColor(grayscale2_uint8, cv2.COLOR_BGR2RGB)
plt.imshow(grayscale2_rgb)
plt.axis("off")
```
