```python
from PIL import Image
import matplotlib.pyplot as plt
```

```python
# https://www.engineeringmadeeasypro.com/2018/07/Historical-Monuments-Famous-Landmarks-and-their-Countries.html
baseImg = Image.open("image.jpg")
# https://www.clker.com/cliparts/P/W/g/2/u/s/grey-water-drop-md.png
# Edited with https://onlinepngtools.com/change-png-opacity
watermark = Image.open("watermark.png")
```

```python
plt.imshow(baseImg)
```

```python
plt.imshow(watermark)
```

```python
simpleWatermarkImg = baseImg.copy()
simpleWatermarkImg.paste(watermark, (0,0), mask=watermark)
```

```python
plt.imshow(simpleWatermarkImg)
```

```python
print(baseImg.size)
print(watermark.size)

xLim = baseImg.size[0] // watermark.size[0]
yLim = baseImg.size[1] // watermark.size[1]
print(xLim)
print(yLim)
```

```python
detailedWatermarkImg = baseImg.copy()
x = 0
y = 0
for _ in range(0, xLim):
    for _  in range(0,yLim):
        detailedWatermarkImg.paste(watermark, (x,y), mask=watermark)
        y += watermark.size[1]
    x += watermark.size[0]
    y = 0
```

```python
plt.imshow(detailedWatermarkImg)
```
